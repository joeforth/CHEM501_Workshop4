## 2. Creating Tables and Inserting Data

In this section, we will define tables for compounds, experiments, and researchers. After creating the tables, we'll insert some sample data to work with.

### 2.1 Creating Tables

````python
%%sql

-- Create Compounds table
CREATE TABLE IF NOT EXISTS Compounds (
    CompoundID INTEGER PRIMARY KEY,
    Name TEXT NOT NULL,
    Formula TEXT,
    MolecularWeight REAL
);

-- Create Researchers table
CREATE TABLE IF NOT EXISTS Researchers (
    ResearcherID INTEGER PRIMARY KEY,
    Name TEXT,
    Department TEXT
);

-- Create Experiments table
CREATE TABLE IF NOT EXISTS Experiments (
    ExperimentID INTEGER PRIMARY KEY,
    CompoundID INTEGER,
    ResearcherID INTEGER,
    Date TEXT,
    Result TEXT,
    FOREIGN KEY (CompoundID) REFERENCES Compounds(CompoundID),
    FOREIGN KEY (ResearcherID) REFERENCES Researchers(ResearcherID)
);



````

## 4. Joining Tables

You can use the JOIN command to combine rows from two or more tables based on a related column between them. 

Let's explore how to join related data from two tables - Compounds and Experiments.

First, we'll remind ourselves what's in each of those tables:

````python
%%sql
SELECT * FROM Compounds;
````

````python
%%sql
SELECT * FROM Experiments;
````

Let's create a new table by joining two tables:

````python
%%sql
SELECT e.ExperimentID, c.Name AS CompoundName, e.Date, e.Result
FROM Experiments e
INNER JOIN Compounds c ON e.CompoundID = c.CompoundID;
````

### 5.2 - GROUP BY and HAVING

You can group data and filter groups by using the `HAVING` clause. You can combine these with JOIN operations to further filter your data.

There are many types of JOIN operation - below we use a LEFT JOIN.

````python
%%sql
SELECT c.Name AS CompoundName, COUNT(e.ExperimentID) AS ExperimentCount
FROM Compounds c
LEFT JOIN Experiments e ON c.CompoundID = e.CompoundID
GROUP BY c.Name
HAVING COUNT(e.ExperimentID) > 0;
````

Like you did with the INNER JOIN, use the internet to understand, line-by-line, what the above code is doing.

### 5.3 - Subqueries

Use subqueries for more complex data retrieval.

````python
%%sql
SELECT Name, Formula FROM Compounds
WHERE CompoundID IN (
    SELECT CompoundID FROM Experiments
    GROUP BY CompoundID
    HAVING COUNT(ExperimentID) > 0
);
````
